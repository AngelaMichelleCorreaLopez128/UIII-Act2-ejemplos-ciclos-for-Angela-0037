#### Ejemplos de Ciclo for

```python

# Ejemplo 1: Cálculo de Nómina Semanal
empleados = int(input("¿A cuántos empleados les calcularás el sueldo?: "))

for i in range(1, empleados + 1):
    horas = float(input(f"Horas trabajadas por el empleado {i}: "))
    pago_hora = 25.50  # Tarifa fija por hora

    # OPERADOR ARITMÉTICO (*): Multiplicamos las horas por el pago
    sueldo = horas * pago_hora

    # EXPLICACIÓN: El ciclo multiplica el esfuerzo (horas) por la tarifa
    # para obtener el sueldo bruto de cada trabajador de forma individual.
    print(f"El sueldo total del empleado {i} es: ${sueldo}")
```

```python
# Ejemplo 2: Sensor de Peso en Producción
piezas = int(input("¿Cuántas piezas pasarán por la báscula?: "))

for i in range(1, piezas + 1):
    peso = float(input(f"Peso de la pieza {i} en gramos: "))

    # OPERADOR RELACIONAL (>= y <=): Verificamos si está en el rango permitido
    if peso >= 450 and peso <= 550:
        # EXPLICACIÓN: Si el peso está entre 450 y 550 gramos, la pieza es válida.
        print(f"Pieza {i}: CALIDAD ACEPTADA.")
    else:
        print(f"Pieza {i}: RECHAZADA (Peso fuera de rango).")
```

```python
# Ejemplo 3: Sistema de Descuentos en Tienda
clientes = int(input("¿Cuántos clientes hay en caja?: "))

for i in range(1, clientes + 1):
    codigo = input(f"Cliente {i}, ¿tiene código de descuento? (si/no): ").lower()
    nivel_vip = int(input("Nivel de membresía (1 al 5): "))

    # OPERADOR LÓGICO (or) y RELACIONAL (>)
    if codigo == "si" or nivel_vip > 3:
        # EXPLICACIÓN: El cliente obtiene descuento si tiene el código
        # O si su nivel de membresía es alto (mayor a 3).
        print("APLICAR DESCUENTO: 20% de rebaja.")
    else:
        print("PRECIO NORMAL: No aplica beneficios.")
```

```python
# Ejemplo 4: División de Equipos de Fútbol
jugadores = int(input("¿Cuántos jugadores se inscribieron?: "))

for i in range(1, jugadores + 1):
    nombre = input(f"Nombre del jugador {i}: ")

    # OPERADOR ARITMÉTICO (%): Residuo para alternar grupos
    if i % 2 == 0:
        # EXPLICACIÓN: Si el número de inscripción es par, va al equipo azul.
        print(f"{nombre} asignado al EQUIPO AZUL.")
    else:
        # EXPLICACIÓN: Si el número es impar, va al equipo rojo.
        print(f"{nombre} asignado al EQUIPO ROJO.")
```

```python
# Ejemplo 5: Simulador de Laboratorio (Bacterias)
horas = int(input("¿Cuántas horas durará la observación?: "))
poblacion_inicial = 10

for h in range(horas + 1):
    # OPERADOR ARITMÉTICO (**): Elevamos 2 a la potencia de la hora actual
    # Esto simula que las bacterias se duplican cada hora.
    poblacion_actual = poblacion_inicial * (2 ** h)

    # EXPLICACIÓN: En cada vuelta (hora), el exponente aumenta,
    # haciendo que el crecimiento sea cada vez más rápido (exponencial).
    print(f"En la hora {h} la población es de: {poblacion_actual} bacterias.")
```
